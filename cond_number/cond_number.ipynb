```python
import glob, os
import numpy as np
```

```python


def read_matrix(path):
    """Read a labelled matrix text file -> numpy array (numeric block only)."""
    with open(path) as f:
        lines = [ln.strip() for ln in f
                 if ln.strip() and not ln.lstrip().startswith("#")]
    # lines[0] is the "gen_bin ..." header -> skip it; skip first token of each row
    rows = [[float(x) for x in ln.split()[1:]] for ln in lines[1:]]
    return np.array(rows, dtype=float)

def condition_numbers(folder="/work/niharrin/devel/RCR_Matrix/cond_number/matrices",
                      pattern="*", p=2):
    paths = sorted(glob.glob(os.path.join(folder, pattern)))
    paths = [p_ for p_ in paths if os.path.isfile(p_)]
    if not paths:
        print(f"No files matching '{pattern}' found in {folder}")
        return {}

    results = {}
    width = max(len(os.path.basename(p_)) for p_ in paths)
    for path in paths:
        name = os.path.basename(path)
        try:
            M = read_matrix(path)
            if M.ndim != 2 or M.shape[0] != M.shape[1]:
                print(f"{name:<{width}}   not square: shape {M.shape}")
                continue
            cond = np.linalg.cond(M, p=p)
            results[name] = cond
            print(f"{name:<{width}}   {cond:.6}")
        except Exception as exc:
            print(f"{name:<{width}}   error: {exc}")
    return results
```
